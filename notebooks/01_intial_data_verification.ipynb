Actual API records check clock-change days

```python
start_london = pd.Timestamp('2024-03-31 00:00', tz='Europe/London')
end_london = pd.Timestamp('2024-04-01 00:00', tz='Europe/London')

start_utc = start_london.tz_convert('UTC')
end_utc = end_london.tz_convert('UTC')
```

```python
start_utc, end_utc
```

```python
params = {
    'from': start_utc.isoformat(),
    'to': end_utc.isoformat(),
    'format': 'json'
}

print(params)
```

```python
spring_api = get_sample_data(
    endpoint=DATA_ENDPOINT,
    params=params
)

spring_api = pd.DataFrame(spring_api)
spring_api['startTime'] = pd.to_datetime(spring_api['startTime'], utc=True)
spring_api = spring_api.loc[(spring_api['startTime'] >= start_utc) & (spring_api['startTime'] < end_utc)].sort_values(by='startTime')
```

```python
spring_api.shape
```

```python
start_london = pd.Timestamp('2024-10-27 00:00', tz='Europe/London')
end_london = pd.Timestamp('2024-10-28 00:00', tz='Europe/London')

start_utc = start_london.tz_convert('UTC')
end_utc = end_london.tz_convert('UTC')
```

```python
start_utc, end_utc
```

```python
params = {
    'from': start_utc.isoformat(),
    'to': end_utc.isoformat(),
    'format': 'json'    
}

autumn_api = get_sample_data(
    endpoint=DATA_ENDPOINT,
    params=params
)
```

```python
autumn_api = pd.DataFrame(autumn_api)
autumn_api['startTime'] = pd.to_datetime(autumn_api['startTime'], utc=True)
autumn_api = autumn_api.loc[(autumn_api['startTime'] >= start_utc) & (autumn_api['startTime'] < end_utc)].sort_values(by='startTime')
```
